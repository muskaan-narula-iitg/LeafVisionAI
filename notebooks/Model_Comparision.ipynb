```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

comparison_df = pd.DataFrame([
    {
        
  "model_name": "EfficientNetB0",
  "validation_accuracy": 0.9899840950965881,
  "validation_loss": 0.12583321332931519,
  "total_parameters": 4392393,
  "model_size_mb": 31.5,
  "batch_size": 32,
  "batch_inference_seconds": 2.2062,
  "seconds_per_image": 0.06894

    },
    {
        
  "model_name": "ResNet50",
  "validation_accuracy": 0.9878215193748474,
  "validation_loss": 0.13927188515663147,
  "total_parameters": 24130214,
  "model_size_mb": 206.9,
  "batch_size": 32,
  "batch_inference_seconds": 1.8697,
  "seconds_per_image": 0.05843
}
    
])

comparison_df = comparison_df.sort_values(
    "validation_accuracy",
    ascending=False
)

display(comparison_df.style.format({
    "validation_accuracy": "{:.2%}",
    "validation_loss": "{:.4f}",
    "model_size_mb": "{:.2f} MB",
    "seconds_per_image": "{:.5f} seconds"
}))

sns.barplot(
    data=comparison_df,
    x="model_name",
    y="validation_accuracy",
    hue="model_name",
    legend=False
)

plt.ylim(0, 1)
plt.title("Validation Accuracy Comparison")
plt.show()

print("Recommended model:", comparison_df.iloc[0]["model_name"])
```

The performance of the two transfer learning models, EfficientNetB0 and ResNet50, was evaluated using validation accuracy, validation loss, model size, number of trainable parameters, and inference time. EfficientNetB0 achieved the highest validation accuracy of 98.998% with a validation loss of 0.1258, outperforming ResNet50, which achieved 98.782% validation accuracy with a validation loss of 0.1393. In addition to its superior predictive performance, EfficientNetB0 required only 4.39 million parameters and occupied approximately 31.5 MB of storage, making it significantly more lightweight than ResNet50, which required 24.13 million parameters and 206.9 MB of storage. Although ResNet50 exhibited a slightly faster inference time (0.05843 seconds per image) compared to EfficientNetB0 (0.06894 seconds per image), the difference was relatively small. Considering its higher classification accuracy, lower validation loss, substantially smaller model size, and lower parameter count, EfficientNetB0 provides a better balance between performance and computational efficiency. Therefore, EfficientNetB0 was selected as the final model for the proposed plant leaf disease detection system.
